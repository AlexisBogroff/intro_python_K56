## Quelques révisions sur les fonctions

Lors de l'appel (utilisation) d'une fonction, l'ordre compte ! Sauf si on a explicité l'ensemble des arguments

```python
def f(prenom, travail):
    return f"{prenom} is a {travail}"

print(f("aleixs", "teacher"))
print(f("teacher", "aleixs"))
print(f(travail="teacher", prenom="aleixs"))
print(f(prenom="aleixs", travail="teacher"))
```

Lors de la définition (création) de la fonction, les arguments sans valeur par défaut doivent être inscrits devant les arguments avec valeur par defaut

```python
def f(prenom, travail="teacher"):
    return f"{prenom} is a {travail}"

f("alexis")
```

```python
# Donc ici ce n'est pas légal !
def f(travail="teacher", prenom):
    return f"{prenom} is a {travail}"

f("alexis")
```

```python
# Puisqu'on a uniquement des arguments avec des valeurs par défaut il n'est plus question d'ordre
def f(travail="teacher", prenom="alexis"):
    return f"{prenom} is a {travail}"

print(f("alexis"))
print(f("teacher", "alexis"))
print(f("teacher", "alexis"))
print(f("teacher", "alexis"))
print(f("teacher", "alexis"))
print(f("teacher", "alexis"))
```
