## 1. What Are Classes and Objects?
A **class** is a blueprint. An **object** is an instance of that blueprint.

**Example:** `Car` is a class → my BMW is an object.

Classes let you group **data** (attributes) and **behaviors** (methods).

## 2. Defining Your First Class
A class is defined with the keyword `class`.
The `__init__` method runs whenever you create a new object.

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

my_car = Car("BMW", "M3")
my_car.brand, my_car.model
```

## 3. Methods: Adding Behavior
Methods are functions defined inside a class.

`self` refers to the instance calling the method.

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def full_name(self):
        return f"{self.brand} {self.model}"

car = Car("Tesla", "Model 3")
car.full_name()
```

## 4. Updating Attributes
Objects can update their internal state.

```python
car = Car("Peugeot", "208")
car.model = "308"  # update attribute
car.full_name()
```

## 5. Example: A Student Class

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def is_passing(self):
        return self.grade >= 10

s1 = Student("Alex", 14)
s2 = Student("Mike", 18)
s1.is_passing()
s2.is_passing()

print(f"Average grade: {(s1.grade + s2.grade)/2}")
```

## 6. Operations

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    
    def __add__(self, other_student):
        return self.grade + other_student.grade


s1 = Student("Alex", 14)
s2 = Student("Mike", 18)

print(f"Average grade: {(s1 + s2)/2}")
```
