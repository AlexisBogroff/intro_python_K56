## Format strings<a class=anchor id=format_strings></a>
Printing both text and variables: sometimes you want some text to go along the value you are printing. Here are 3 common ways to do it.

````python
val1 = 222
val2 = 333
````

Simple format V0:

````python
print('result1:', val1, ', result2:', val2)
````

Advanced format V1: `.format()`

````python
print('result1: {}, result2: {}'.format(val1, val2))
print('result1: {0}, result2: {1}'.format(val1, val2))
print('result1: {num1}, result2: {num2}'.format(num1=val1, num2=val2))
````

Advanced format V2 (since Python 3.6): `f-string`

````python
print(f'result1: {val1}, result2: {val2}')
````

### Lists operations

A list is an iterable (we can extract every element iteratively). There exist 4 very useful methods to work with them.

#### 1. Extracting one or multiple elements

````python
l = ['a', 'b', 'c']

# Q1: Display the first element
print('q1:', l[0])
# Q2: Display the second element
print('q2:', l[1])
# Q3: Display the third element
print('q3:', l[2])

# Q4: Display the last element
print('q4:', l[-1])
# Q5: Display the before last element
print('q5:', l[-2])

# Q6.1: Display the first two elements
print('q6.1:', l[0:2])
# Q6.2: Same, but more canonical
print('q6.2:', l[:2])

# Q7.1: Display the first three elements
print('q7.1:', l[0:3])
# Q7.2: Same, but more canonical
print('q7.2:', l[:3])

# Q8: Display the last two elements
print('q8:', l[-2:])
# Q9: Display the last three elements
print('q8:', l[-3:])

# What is the type of the object returned when only a single element is extracted?
# What is the type of the object returned when multiple elements are extracted at the same time?
````

#### Chaining operations

We can actually use a combination of functions. To understand how it works you should mentally replace each operation by its result. Each operation is applied on the result of the previous set of operations

````python
# Extract the first element of the list contained in the key 'surname'
# First, understand how to extract the list contained in the dictionnary under the key 'surname'
print(dico['surname'])

# Second, since we know how to obtain the list, we can simply use the method to extract elements of a list
# Using the square brackets and the number indicating the position of the element to extract (here 0 for the first)
print(dico['surname'][0])
````

````python
# Extract the second element of the list contained in the key 'surname'
print(dico['surname'][1])

# Extract the last element of the list contained in the key 'age'
print(dico['age'][-1])
````

````python
# Execute the function print contained in the dictionary under the key 'objects'

# First, extract the function print
print(dico['objects']) 

# Second, execute the function since dico['objects'] is equivalent to print
# Thus make a standard use of it with parenthesis containing a text print('ok')
dico['objects']('ok')
````

````python
# Let's use some functions from the standard library

# Join all surnames and separate them with a dash
print('-'.join(dico['surname']))

# Then put them in capital letters
print('-'.join(dico['surname']).upper())

# Then extract the third and fourth letters
print('-'.join(dico['surname']).upper()[3:5])

# Then transform them back to lower letters
print('-'.join(dico['surname']).upper()[3:5].lower())

# Instead of extracting them, keep the whole suite of letters with the same modifications than above
# This could be done with a cleaner and code, but we keep it complex to make a point (see in next cell)
'-'.join(dico['surname']).upper()[:3] + '-'.join(dico['surname'])[3:5] + '-'.join(dico['surname']).upper()[5:]

# Do the same with a shorter one-liner
'-'.join(dico['surname']).upper()[:3] + '-'.join(dico['surname'])[3:5] + '-'.join(dico['surname']).upper()[5:]
````

````python
# Okkkkkk we get it! Its possible to compose functions indefinitely, but its not digestible anymore.
# Let's use variables for intermediate steps, and appreciate how its way clearer
# Even though it requires more lines of code, its easier to understand and maintain (so its definitely prefered)
surnames_combined = '-'.join(dico['surname'])
surnames_combined_upper = surnames_combined.upper()

middle_letters = surnames_combined[3:5]

start_letters = surnames_combined_upper[:3]
end_letters = surnames_combined_upper[5:]

full_combination = start_letters + middle_letters + end_letters
full_combination
````

````python
# That was clearer but very verbosy!
# Depending on the use case we should try to use the best balance between short code vs long and readable
# V1
surnames_combined = "-".join(dico["surname"])
full_combination = surnames_combined.upper()[:3] + surnames_combined[3:5] + surnames_combined.upper()[5:]
print(full_combination)

# V2
surnames_combined = "-".join(dico["surname"]).upper()
full_combination = surnames_combined[:3] + surnames_combined[3:5].lower() + surnames_combined[5:]
print(full_combination)
````

````python
# Eventually we can accept to name the variable in a more compact way if it is not used lated in the code (here 's')
# Please don't name variables like this if used later in the code, because you would not understand what 's' means
# which would force you to go back to this portion of code to understand what 's' means before using it.
# Ideally we want the variable name to be self-explanatory and obvious enough to avoid having to go back to the portion
# of code where it was defined (imagine doing this for each line of code, it would be a big time waste)
s = "-".join(dico["surname"])
full_combination = s.upper()[:3] + s[3:5] + s.upper()[5:]
full_combination
````

````python
# We can even make use of more advanced coding practices, but if its not clearer, its better to use the long verbosy version
# We'll explore the lambda version later in this notebook
(lambda s: s[:3].upper() + s[3:5] + s[5:].upper())("-".join(dico["surname"]))
````

## Control flow structures<a class=anchor id=control_structures></a>

For loop: is used to work on **all** the elements of an iterable (an iterable is often a list or a dictionnary)

### Working on lists

````python
# With simple list
# Here I use the letter 'i', but this is actually a variable bein defined at each iteration
# which means you can name it as you like. In most cases you prefer using meaningful names 
# that help understand the code (e.g. 'temperature' instead of 'i')

# Inside the loop we decide to simply do a print here, but any suite of operation can be conducted
# its possible to have hundred lines of code within the for loop statement or even more
for i in [20, 17, 30]:
    print(i)

i = 20
print(i)
i = 17
print(i)
i = 30
print(i)
````

````python
# Let's do a more informative print
for temperature in [20, 17, 30]:
    print(f'The current temperature is {temperature}°C')
````

````python
# Very often the list is has already been defined
# See how the spelling helps understand whats happening for temperature in temperatures (with a 's')
temperatures = [20, 17, 30]

for temperature in temperatures:
    print(f'The current temperature is {temperature}°C')
````

### Working on dictionaries

````python
# For dictionnaries, by default the loop iterates over the keys only (not the values)
for key in dico:
    print(key)
````

````python
# Here are two methods to get the values instead

# Method 1: iterate directly on the values
for val in dico.values():
    print(val)

# Method 2: iterate on the keys that you can use to get the values
for key in dico:
    print(dico[key])
````

````python
# To iterate over both keys and values you can use the .items method
for combo in dico.items():
    print(combo)
````

````python
# Another way to do the same (cleaner) is done by definined two variables in the for loop
# (2 exactly because we expect two values for each iteration of the for loop because a dicitonary is composed of PAIRS of keys and values)
for key, val in dico.items():
    print(f'The key is: {key}, and the val is: {val}')
````

````python
# Let's do the same with a different output format
# '\n' means carriage return (skip / jump a line)
for key, val in dico.items():
    print(f'The key is: {key}\nAnd the val is: {val}\n')
````

````python
# Let's iterate over all the values contained in age
# We actually extract the list of ages and iterate over the list instead of the dicitonary
for age in dico['age']:
    print(f'The current age is {age} years old.')
````

## Operators<a class=anchor id=operators></a>

### Comparisons

The following are **operators of comparison**. They enable comparisons between objects of the same type (by default).

The result is a boolean (True or False). The result is often used in conditional structures (if, elif, else).

````python
# Each object (integers, strings, lists, etc.) has its own operator
# which means that "greater than" when comparing integers 
# has a different meaning than "greater than" for comparing lists
# ps: you will see later how to define operators for your own objects
print(1 > 2)
print(1 < 2)
print(1 < 1)
print(1 >= 2)
print(1 <= 2)
print(1 <= 1)
print(1 == 2)
print(1 != 2)
print(1 == 1)
print(1 != 1)
print('a' < 'b')
print('a' > 'b')
print([1] < [2, 3])
print([1] > [2, 3])
````

**Contains** operator: available in iterables, not in numbers.

````python
print(1 in [1, 4, 2])
print(1 in [4, 2])
print([1] in [1, 4, 2])
print([1] in [[1], 4, 2])
print('a' in 'oisj')
print('a' in 'oiasj')
````

Check if element is (True, False) or if it even exists

````python
print(True is True)
print(True is False)

print([1] is None)
x = 12
print(x is None)
x = None
print(x is None)

import numpy as np
print(x is np.nan)
````

**Negation** operator: not

````python
print(1 == 1)
print(not 1 == 1)
print(1 != 1)
print(not 1 != 1)
````

## Functions<a class=anchor id=functions></a>

Functions can:
- take no, to many arguments
- arguments can be "Positional" or "Keyword" arguments
- return nothing (None) or anything (to many things)
- synonyms: arguments / parameters / inputs

One must:
- Define the function
- Call the function

Simplest form:
- No argument required
- No return

````python
# Define the function
def say_something():
    print("Something")

# Call the function
say_something()
````

Function with:
- an argument
- no return

````python
def say_my_name(name):
    print(name)

say_my_name("truc")
````

Function with:
- an argument
- a return

````python
def square_print(x):
    print(x**2)

def square_return(x):
    return x**2

# result = square_return(4)
result = square_print(4)
result
````

Function with:
- multiple arguments
- a return

````python
def add(a, b, c):
    return a + b + c

result = add(4, 2, 9)
result
````

Function with:
- no argument
- multiple returns

````python
def return_many_things():
    return 'alexis', 'bogroff', 'data'

return_many_things()
````

````python
# Get the result in separate variables
r_1, r_2, r_3 = return_many_things()
print(r_1)
print(r_2)
print(r_3)
````

Function with:
- a keyword argument
    - is thus optional
    - must be positioned after the positional arguments
- no return

````python
def say_what_you_doing(name, course='data'):
    print(f"{name} doing {course}")

say_what_you_doing("Alexis")
say_what_you_doing("Alexis", "writing the course")
````

Function with:
- a (positional) argument awaiting a function
- no return

````python
def complex_fct(func):
    print("This function will say")
    func()

complex_fct(say_something)
````

## Lists comprehension<a class=anchor id=lists_comprehension></a>

The standard way to create a list from a loop

````python
l_loop = []
for i in range(10000):
    l_loop.append(i**2)
````

The efficient way, list comprehension

Advantage:
- one liner
- can be easier to read in specific situations

Disadvantages:
- less readable (in most situations when the logic / calculus is a bit complex)
- not even possible for very complex operations within the loop

````python
l_comprehension = [i+10 for i in [1, 4, 2, 9, 4]]
l_comprehension
````

````python
# Generate complex elements in the list
l_comprehension = [('a', i, i+10) for i in [1, 4, 2, 9, 4]]
l_comprehension
````

````python
# Use range to create a long list
l_comprehension = [i**2 for i in range(10000)]
l_comprehension[:5]
````

Use a condition

````python
l_comprehension = [i**2 for i in range(10000) if i%10==0]
l_comprehension[:5]
````

Use a more complex condition (it requires to write the if at the beginning)

````python
l_comprehension = [i**2 if i%2==0 else i for i in range(10000)]
l_comprehension[:5]
````

## Objects<a class=anchor id=objects></a>

### Create simple class

````python
class Person:

    def __init__(self, age, prenom):
        self.age = age
        self.prenom = prenom
    
    def marche(self):
        print("je marche")
    
    def arret(self):
        print("je m'arrête")

    def parle(self):
        print("je parle")
````

````python
person_1 = Person(12, "alexis")
person_2 = Person(24, "truc")
````

````python
# On peut accéder aux attributs de la même manière pour chaque instance (chaque personne)
# C'est pratique !
print(person_1.age)
print(person_2.age)

print(person_1.prenom)
print(person_2.prenom)
````

````python
# Idem pour les méthodes (fonctions)
person_1.parle()
person_2.parle()
````

### Instanciate with an argument

````python
class Truc:

    # Define instanciator (init)
    def __init__(self, name, age):
        self.age = age
        self.name = name

    def present(self):
        print(f"My name is: {self.name}, "
              f"I'm {self.age} years old")

    def dog_age(self):
        return self.age * 7
````

````python
truc_1 = Truc('Yuko', 7)
truc_2 = Truc('Mila', 13)
````

````python
print(f"{truc_1.name} is {truc_1.dog_age()} old")
print(f"{truc_2.name} is {truc_2.dog_age()} old")
````
